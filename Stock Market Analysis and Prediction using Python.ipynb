# ***Part 1 Data Import and Visualization of Stocks***

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
import datetime as dt

s = dt.date(2022,1,1)
e = dt.date(2025,11,30)
tickers = ["HCA", "GIS","SRE"] 

df = yf.download(tickers,s,e)["Close"]
print(df.head())

# Visualization of Stocks
plt.figure(figsize=(12, 6))
for ticker in tickers:
    plt.plot(df.index, df[ticker], label=ticker)
plt.title("Daily Stock Prices (Last 3 Years)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.show()
```

# ***Part 2 ARIMA Models***

```python
# ARIMA Model on HCA Healthcare Inc Stock Prices
from statsmodels.tsa.ar_model import AutoReg
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error
import statsmodels.api as sm
from math import sqrt
from sklearn.metrics import mean_absolute_error, mean_squared_error

data = df["HCA"].dropna()

train_ratio = 0.8
train = data[:int(train_ratio*len(data))]
test = data[int(train_ratio*len(data)):] 

# Train model (ARMA = ARIMA with no trend)
model = ARIMA(train, order=(10,0,1)).fit() 

# Predict
pred = model.predict(start=len(train), end=len(data)-1) 

# Errors
mae = mean_absolute_error(test, pred)
mse = mean_squared_error(test, pred)
rmse = sqrt(mse)
 
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)

# Visualization
plt.figure(figsize=(10,6)) 
# Plot actual test values
plt.plot(test.index, test.values, label='Actual Series') 
# Plot predicted values
plt.plot(test.index, pred.values, label='Predicted Series') 
# Title and labels (recommended)
plt.title("Actual vs Predicted Stock Price")
plt.xlabel("Index")
plt.ylabel("Stock Price") 
plt.legend()
plt.show()
```

# ***Part 3A Co-integration***

```python
from statsmodels.tsa.stattools import adfuller, coint

def adf_test(series, name=''):
    result = adfuller(series)
    print(f"ADF Test for {name}:")
    print(f" Test Statistic = {result[0]:.4f}")
    print(f" p-value = {result[1]:.4f}")
    print('  => Non-stationary' if result[1] > 0.05 else '  => Stationary')
    print()
adf_test(df["HCA"], "HCA Healthcare Inc")
adf_test(df["GIS"], "Genral Mills Inc")

"If p-value < 0.05 → HCA Healthcare Inc and Genral Mills INC are cointegrated (long-run relation exists)."
"If not → they just happen to trend similarly."
score, pvalue, _ = coint(df["HCA"], df["GIS"])
print(f"Engle-Granger Cointegration Test p-value: {pvalue:.4f}")
if pvalue < 0.05:
    print("These stocks are Co-integrated.")
else:
    print("These stocks are not Co-integrated.")
```

```python
# Polynomial Regression
X = data[["Years_Experience"]]
y = data["Salary"]

# Transform features to polynomial
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")

# Visualization
plt.figure(figsize=(8, 5))
plt.scatter(data["Years_Experience"], data["Salary"], color="blue")

# Smooth curve plotting
X_range = np.linspace(data["Years_Experience"].min(), data["Years_Experience"].max(), 100).reshape(-1, 1)
y_range_pred = model.predict(poly.transform(X_range))
plt.plot(X_range, y_range_pred, color="green", label="Polynomial Fit")
plt.title("Polynomial Regression")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```
